```python
matiere = [f"R{i}" for i in range(101,116)]+[f"SAE{i}"for i in range(11,17)]
list_notes = [i for i in range(21)]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

dico_notes = {cle : valeur for cle, valeur in zip (matiere,list_notes)}
dico_coef1 = {cle : valeur for cle, valeur in zip (matiere,coef1)}
dico_coef2 = {cle : valeur for cle, valeur in zip (matiere,coef2)}
dico_coef3 = {cle : valeur for cle, valeur in zip (matiere,coef3)}

def dico(list_valeurs):
    return {cle : valeur for cle, valeur in zip (matiere,list_valeurs)}

dico_notes = dico(list_notes)
dico_coef1 = dico(coef1)
dico_coef2 = dico(coef2)
dico_coef3 = dico(coef3)
```

```python
def moy(dico_coef):
    somme = 0
    total_coef = 0
    
    for element in matiere :
        somme += dico_coef[element]*dico_notes[element] 
        total_coef += dico_coef[element]
    return round(somme / total_coef,3)

print(f"La moyenne de l'UE1 est {moy(dico_coef1)}, celle de l'UE2 est {moy(dico_coef2)} et celle de l'UE3 est {moy(dico_coef3)}")
```
