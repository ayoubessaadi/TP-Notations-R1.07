```python
from tkinter import *
from tkinter import ttk

fen = Tk()
fen.geometry("800x600")
fen.title("Travail sur l'interface graphique")

frm = ttk.Frame(fen)
frm.grid()

objet1 = ttk.Label(frm, text = "Ceci est le premier texte")
objet1.grid(column = 0, row = 0, padx = 60, pady = 20)

objet2 = ttk.Entry(frm)
objet2.grid(column = 1, row = 0)

# on defie la fonction et on affiche le résultat dans la fenetre
def recup():
    objet3 = ttk.Label(frm, text = float(objet2.get())+3)
    objet3.grid(column = 2, row = 3)
    
# pour crée le boutton valider
btn1 = ttk.Button(frm, text = "Valider", command = recup)
btn1.grid(column = 2, row = 2)

fen.mainloop()
```

```python
from tkinter import *
from tkinter import ttk
import numpy as np

#création de la fenêtre
fen = Tk()
fen.title("Exercice")

#création d'un cadre dans la fenêtre et d'une grille 
fram = ttk.Frame(fen)
fram.grid()

#création des matières

matiere = [f"R{i}" for i in range (101, 116)]+[f"SAE{i}" for i in range (11,17)] #reprend tout les R ainsi que les SAE en une ligne

#création des listes de coefficients
saisies=[]
coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]


for i, matiere in enumerate(matiere) : 
   mat = ttk.Label(fram, text = matiere)
   mat.grid(column = 0 , row = i+1, padx = 20, pady = 10)
   data = ttk.Entry(fram)
   data.grid(column = 1, row = i+1)
   saisies.append(data)

def recup():
   notes = [float(data.get()) for data in saisies]

   UE1 = ttk.Label(fram, text = "Moyenne de l'UE1")
   UE1.grid(column = 2, row = 3, pady = 20)
   moy1 = ttk.Label(fram, text = round(np.average(notes, weights = coef1),3), relief = GROOVE)
   moy1.grid(column = 3, row = 3, pady = 20)

   UE2 = ttk.Label(fram, text = "Moyenne de l'UE2")
   UE2.grid(column = 2, row = 5, pady = 20)
   moy2 = ttk.Label(fram, text = round(np.average(notes, weights = coef2),3), relief = GROOVE)
   moy2.grid(column = 3, row = 5, pady = 20)

   UE3 = ttk.Label(fram, text = "Moyenne de l'UE3")
   UE3.grid(column = 2, row = 7, pady = 20)
   moy3 = ttk.Label(fram, text = round(np.average(notes, weights = coef3),3), relief = GROOVE)
   moy3.grid(column = 3, row = 7, pady = 20)
   

#bouton positionné à droite
btn1 = Button(fram, text="Valider la saisie", command=recup)
btn1.grid(column=3, row=1, padx=15)

#maintient la fenêtre ouverte
fen.mainloop()s
```
